```python
# Célula 1: Importações e Configurações
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração para gráficos ficarem bonitos no notebook
sns.set_theme(style="whitegrid")
%matplotlib inline

# Carregando os dados (ajuste o caminho se necessário)
# O dataset 'day.csv' é geralmente o mais usado para predição diária
df = pd.read_csv('../data/raw/day.csv')

# Visualizando as primeiras linhas para garantir que carregou certo
display(df.head())

# Verificando informações básicas (tipos e nulos)
print("\nInformações do Dataset:")
df.info()
```

```python
# Célula 2: Análise de Correlação e Vazamento de Dados

# 1. Calcular a matriz de correlação
corr_matrix = df.corr(numeric_only=True)

# 2. Plotar o Heatmap (Mapa de Calor)
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Matriz de Correlação - Atenção para 'casual' e 'registered'")
plt.show()

# 3. Plotar a relação Temperatura vs Total de Aluguéis
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='temp', y='cnt', hue='season', palette='viridis')
plt.title("Relação entre Temperatura e Aluguéis (cnt)")
plt.xlabel("Temperatura Normalizada")
plt.ylabel("Total de Aluguéis")
plt.show()
```
